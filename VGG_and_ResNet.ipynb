# Dataset of imagenet labels

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = 'https://deeplearning.cms.waikato.ac.nz/user-guide/class-maps/IMAGENET/'

req = requests.get(url)
html_doc = req.text
```

```python
html_doc
```

```python
html_bs = BeautifulSoup(html_doc, 'lxml')
html_bs
```

```python
header_html = html_bs.find_all('th')
header_html
```

```python
row_html = html_bs.find_all('tr')
row_html
```

```python
row_html.pop(0)
```

```python
len(row_html)
```

```python
headers = []
for h in header_html:
  headers.append(h.text)

rows = []
for row in row_html:
  row_data = []
  for cell in row.find_all('td'):
    row_data.append(cell.text)
  rows.append(row_data)
```

```python
len(rows)
```

```python
df = pd.DataFrame(rows, columns=headers)
df = df.drop(['Class ID'], axis=1)
df.head()
```

```python
df.to_csv('imagenet_class_name.csv')
```

### VGG16

```python
from keras.applications import VGG16
from keras.applications.vgg16 import preprocess_input, decode_predictions
from keras.preprocessing.image import load_img, img_to_array
from tensorflow import expand_dims
```

```python
vgg_model = VGG16(weights='imagenet')
```

```python
img_path = ['tinca tinca.jpg', 'white shark.jpg', 'african lion.jpg']
for img in img_path:
  image = load_img(img, target_size=[224, 224, 3])
  array_img = img_to_array(image)
  array_img = expand_dims(array_img, axis=0)
  vgg_img = preprocess_input(array_img)
  pred = vgg_model.predict(vgg_img)
  top_acc = 5
  for j in range(top_acc):
    print(decode_predictions(pred, top=top_acc)[0][j][1])
  print('------------------------------------------')
```

### VGG16

```python
from torchvision.models import vgg16
from torchvision import transforms
import torch
from PIL import Image
```

```python
vgg_model = vgg16(pretrained=True)
vgg_model.eval()
```

```python
preprocess = transforms.Compose([transforms.Resize(256),
                                 transforms.CenterCrop(224),
                                 transforms.ToTensor(),
                                 transforms.Normalize(mean=[0.485, 0.456, 0.406],
                                                      std=[0.229, 0.224, 0.225])]
                                )
img = Image.open(img_path[1])
img = preprocess(img)
img = torch.unsqueeze(img, 0)

with torch.no_grad():
  pred = vgg_model(img)

_, indices = torch.sort(pred, descending=True)
print(indices[0][0:5])
```

```python
with open("imagenet_class_name.csv") as f:
  classes = [line.strip() for line in f.readlines()[0:5]]
```

```python
_, index = torch.max(pred, 1)
percentage = torch.nn.functional.softmax(pred, 1)[0] * 100
print(classes[index[0]+1], percentage[index[0]].item())
```

### ResNet

image used in this notebook:

[jay](https://plantura.garden/uk/wp-content/uploads/sites/2/2021/03/jay-perches-on-branch.jpg)

```python
import matplotlib.pyplot as plt
from torchvision.io import read_image
from torchvision.models.quantization import resnet50, ResNet50_QuantizedWeights

img = read_image('jay.jpg')

plt.imshow(img.cpu().permute(1, 2, 0).numpy())
```

```python
weights = ResNet50_QuantizedWeights.DEFAULT
resnet_model = resnet50(weights=weights, quantize=True)

resnet_model.eval()

preprocess = weights.transforms()

img_batch = preprocess(img).unsqueeze(0)

plt.imshow(img_batch.squeeze(0).cpu().permute(1, 2, 0).numpy())
```

```python
prediction = resnet_model(img_batch).squeeze(0).softmax(0)
class_id = prediction.argmax().item()
score = prediction[class_id].item()
category_name = weights.meta['categories'][class_id]
print(f"{category_name} : {score * 100}")
```
